```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
YearsOfExperience = np.random.randint(2, 21, size=500)
intercept = 60000
slope = (200000 - 60000) / (20 - 2)
noise = np.random.normal(0, 10000, size=500)
Salary = intercept + slope *YearsOfExperience + noise
```

```python
data = {
    'YearsOfExperience': YearsOfExperience,
    'Salary': Salary
}
df = pd.DataFrame(data)
df.head()
df.shape
```

```python
df.shape
df.describe()
df.isnull().sum()
df.corr()
```

```python
df.plot(kind='scatter', x='YearsOfExperience', y='Salary')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Salary vs Experience')
plt.show()
```
